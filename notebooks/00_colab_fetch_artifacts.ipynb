# Colab corpus artifact fetch

Run these cells in Colab to fetch the public GitHub LFS parquet artifacts by raw URL, install the analysis dependency group with `uv`, and regenerate the corpus analysis outputs without cloning the repository.

```python
repo = "bru02/epstein"
ref = "autoresearch/preprocess-pipeline-2026-04-17"
root = "/content/epstein"
base_url = f"https://github.com/{repo}/raw/{ref}"
```

```python
%pip install -q uv
```

```python
from pathlib import Path
from urllib.request import urlretrieve

Path(root, "data").mkdir(parents=True, exist_ok=True)

for remote_path in [
    "pyproject.toml",
    "uv.lock",
    "corpus_text_analysis.py",
    "data/fact_emails.parquet",
    "data/bridge_email_people.parquet",
    "data/dim_people.parquet",
]:
    target = Path(root, remote_path)
    target.parent.mkdir(parents=True, exist_ok=True)
    url = f"{base_url}/{remote_path}"
    print(f"fetching {url}")
    urlretrieve(url, target)
    print(target, target.stat().st_size)
```

```python
%cd /content/epstein
!uv sync --only-group corpus-analysis
```

```python
!uv run --no-sync python corpus_text_analysis.py
```
